**Q1 — Pass or Fail Counter**   
20 min  ·  loops, conditionals, comprehensions  
You're given a hardcoded list of exam marks:  
marks = [45, 32, 78, 39, 90, 28, 55, 40, 67, 12]    

Write a script that:
* Loops through the list and prints whether each mark is a Pass (40 or above) or a Fail
* Counts how many passed and how many failed, using simple counters
* Uses a comprehension to build a new list containing only the passing marks
* Prints the final counts and the passing-marks list

```python
#Q.1 Pass or fail counter

marks = [45, 32, 78, 39, 90, 28, 55, 40, 67, 12]

passed = 0 # Created a counter to track a how many students are passed or failed
failed = 0

for num in marks: #used for loop, to check each marks one by one
    if num >= 40: #we used in taht camparision operators.
        print(num, "Pass") # if its 40+ marks it prints 'Pass'
        passed += 1 #increasing the pass counter by 1.
    else:
        print(num, "Fail")
        failed += 1

passing_marks = [num for num in marks if num >= 40] #created a list for only those marks that are 40 or more than that.

print("Total Passed:", passed)
print("Total Failed:", failed)
print("Passing Marks:", passing_marks)
```

**Q3 — Daily Step Tracker**
20 min · random module, loops, math module  
Simulate a fitness tracker recording step counts for 7 days, using the random module.   

* Use random.randint(2000, 12000) to generate each day's step count
* Store all 7 values in a list, printing each one as it's generated (e.g. "Day 1: 8421 steps")
* After the loop, print the highest and lowest day using max() and min()
* Print the average steps, and use math.ceil() to round it up to the nearest whole number

```python
# Q.3  Daily Step Tracker

import random
import math

steps = []

for day in range(1, 7):
    daily_steps = random.randint(2000, 12000) # generating the random no. between 2000 - 12000
    steps.append(daily_steps) #adding step,
    print(f"Day {day}: {daily_steps} steps")

print("Highest:", max(steps)) #Finding highest value
print("Lowest:", min(steps)) #Finding lowest value

average = math.ceil(sum(steps) / len(steps))
print("Average:", average) # calculating the average value
```

**Q4 — Student Class**  
30 min  ·  classes, __init__, type hints, one dunder method   
Design a Student class:   
* __init__(self, name: str, marks: list) — both type-hinted instance attributes
* A method average(self) -> float that returns the average of the student's marks
* __str__ — returns a readable one-line summary, e.g. "Sakshi: average 82.5"
* Then create a hardcoded list of 3 Student objects, print each one (this should use your __str__ automatically), and print the name of whichever student has the highest average.

```python
# Q.4 Student Class

class Student: # Created a student class with name and marks
    def __init__(self, name: str, marks: list):
        self.name: str = name
        self.marks: list = marks

    def average(self) -> float: #Creating the average function
        return sum(self.marks) / len(self.marks) # calculating the average marks

    def __str__(self): #Created a function using dunder method for printing the output name and average
        return f"{self.name}: average {self.average():.1f}"

#Create 3 student objects
student1 = Student("Aaditya", [67, 80, 95])
student2 = Student("Shruti", [88, 90, 93])
student3 = Student("Sanskar", [93, 74, 80])

students = [student1, student2, student3] #Store all the student object in list

for student in students: # Printing every student
    print(student)

highest_student = students[0]

for student in students:
    if student.average() > highest_student.average():
        highest_student = student

print("Highest average:", highest_student.name)
```

Using only what's been shown in class (the random, math, and sys modules — pick any one or combine them):

```python
#Q.5  Import and Use a Module

import random as rd

count = 0
numbers = [1, 2, 3, 4, 5, 6]

while True:
    # Choose a random number from the list
    dice = rd.choice(numbers)

    count += 1
    print("Dice:", dice)

    if dice == 6:
        break

# Generate a random number between 1 and 10
bonus = rd.randint(1, 10)

print("Number of rolls:", count)
print("Bonus number:", bonus)
```
