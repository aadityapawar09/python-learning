```python
#Break
while True:
  print('Enter Your name.')
  name = input('>')
  if name == 'Your Name':
    break #it will use when u want to stop the loop
  print('Thank You')
```

```python
#Continue

for num in range (10):
  if num % 3 == 0:
    continue #it skips the no, and jump to the next.
  print(num)
```

Break vs Continue

Break                             
1. Loop is running
2. break executes
3. loop ends immediately
4. Code after the loop runs
The loop never cheaks its condition again

Continue
1. Loop is running
2. Continue executes
3. Rest of this pass is skipped
4. condition is cheaked again
The loop keeps going - just skips ahead
